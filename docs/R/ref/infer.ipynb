# Construct confidence intervals and perform statistical inference for fitted CGDRO models

```python
infer_cgdro_(
    fit,
    M = 50,
    alpha = 0.05, 
    ...
)
```

## Arguments

**fit**

A list of results from `cgdro_()`.

**M**

Number of Monte Carlo samples used to construct the intervals.
Default is 50.

**alpha**

Significance level for interval construction.
Default is 0.05.


**...**

See [`family`](/R/ref/family/)



## Outputs

**CI**

Confidence intervals for the estimators of interest.
