# Generate predictions from fitted CGDRO models on the target domain

```r
predict_cgdro_(
    fit, 
    X = NULL
)
```

## Arguments

**fit**

A list of results from `cgdro_()`.

**X**

Optional numeric design matrix for the target domain. If NULL, the matrix fit$X0_use defined during fitting will be used.


## Outputs

**pred**

Predicted labels on the target domain.

**pred_proba**

Predicted probabilities on the target domain. Only for `family = 'cls'`.
