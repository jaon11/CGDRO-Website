# Summarize the training and inference results, including group weights, coefficient estimators and confidence intervals

```python
summary_cgdro_(
    fit,
    infer = NULL, 
    ...
) 
```

## Arguments
`summary_cgdro_()` works for `family = 'reg_ld', 'reg_hd', 'cls'`.


**fit**

A list of results from `cgdro_()`.


**infer()**

A list of results from `infer_cgdro_()`.

**...**

See [`family`](/R/ref/family/)



## Outputs

A table of summarized CGDRO aggregated results, including group weights, coefficient estimators and confidence intervals.
